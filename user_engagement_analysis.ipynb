# Análise de Engajamento de Usuários em Plataforma Digital

## 1. Introdução

Este projeto tem como objetivo analisar padrões de engajamento de usuários em uma plataforma digital, utilizando dados simulados. A partir disso, buscamos entender o comportamento de diferentes grupos de usuários e propor melhorias com base em dados.


```
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import warnings
warnings.filterwarnings('ignore')
sns.set(style='whitegrid')
```

```
df = pd.read_csv('user_engagement.csv')
df.head()
```

```
df.info()
df.isnull().sum()

# Garantir que não haja divisão por zero
df['page_views'] = df['page_views'].replace(0, 1)

# Criar coluna de taxa de cliques
df['click_rate'] = (df['clicks'] / df['page_views']).round(2)
df['conversion'] = df['conversion'].astype(int)
```

## 2. Análise Exploratória

```
# Tempo médio de sessão por tipo de dispositivo
sns.boxplot(data=df, x='device_type', y='session_time')
plt.title('Tempo de Sessão por Tipo de Dispositivo')
plt.show()
```

```
# Taxa de conversão por país
sns.barplot(data=df.groupby('country')['conversion'].mean().reset_index(),
            x='country', y='conversion')
plt.title('Taxa de Conversão por País')
plt.show()
```

## 3. Clusterização de Usuários

```
features = df[['session_time', 'clicks', 'conversion', 'click_rate']]

scaler = StandardScaler()
features_scaled = scaler.fit_transform(features)

kmeans = KMeans(n_clusters=3, random_state=42)
df['cluster'] = kmeans.fit_predict(features_scaled)

df.groupby('cluster').mean()
```

## 4. Principais Insights

- Cluster 0: usuários mais engajados e com maior conversão.
- Cluster 2: usuários com engajamento mais baixo.
- Dispositivos Android têm menor tempo médio de sessão → investigar UX.


## 5. Conclusão

A análise permitiu identificar diferentes perfis de usuários e apontar oportunidades de melhoria na experiência da plataforma, especialmente para usuários menos engajados. A clusterização foi útil para segmentar o público com base em comportamento.
